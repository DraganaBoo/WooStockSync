```python
import pandas as pd
import os

folder = r"C:\Users\online1\Desktop\POWER QUERY\PYTON"

# Fajlovi
lager_file = "Lager.xlsx"
wc_file = "wc-product-export.csv"

# Učitavanje fajlova
lager = pd.read_excel(os.path.join(folder, lager_file), engine='openpyxl')
wc = pd.read_csv(os.path.join(folder, wc_file), decimal=',', low_memory=False)

# Pretvori kolone u numeričke tipove
lager["Na zalihama"] = pd.to_numeric(lager["Na zalihama"], errors='coerce')
lager["MP Cena"] = pd.to_numeric(lager["MP Cena"], errors='coerce')
wc["Zalihe"] = pd.to_numeric(wc["Zalihe"], errors='coerce')
wc["Redovna cena"] = pd.to_numeric(wc["Redovna cena"], errors='coerce')

# Spajanje fajlova po šifri
merged = pd.merge(lager, wc, left_on="Broj stavke", right_on="Šifra proizvoda", how="outer")

# Funkcija za preračunavanje
def convert(row):
    if row["Jedinica mere prodaje"] == "M":
        row["Zalihe_konvertovane"] = row["Na zalihama"] / 100
        row["Cena_konvertovana"] = row["MP Cena"] * 100
    elif row["Jedinica mere prodaje"] == "C":
        row["Zalihe_konvertovane"] = row["Na zalihama"] * 100
        row["Cena_konvertovana"] = row["MP Cena"] / 100
    else:
        row["Zalihe_konvertovane"] = row["Na zalihama"]
        row["Cena_konvertovana"] = row["MP Cena"]
    return row

merged = merged.apply(convert, axis=1)

# Sačuvaj finalni Excel fajl
final_file = os.path.join(folder, "Finalni_Webshop.xlsx")
merged[["Broj stavke","Šifra proizvoda","Jedinica mere prodaje","Zalihe_konvertovane","Cena_konvertovana"]].to_excel(final_file, index=False)

print("Gotovo! Finalni fajl je sačuvan kao:", final_file)
```
